## adding data to the database 

```python
def Add_Details():
    

    name = input("Enter the birthday boy/girl's name: ")
    DOB = input("Enter the DOB (YYYY-MM-DD): ")
    Email = input("Enter the email to send an automatic birthday message: ")

    try:
        cursor.execute(
            """
            INSERT INTO B_Boys (name, DOB, Email)
            VALUES (?, ?, ?)
            """,
            (name, DOB, Email)
        )

        conn.commit()

        print("All the required details are stored in the database!")

    except sqlite3.Error as e:
        conn.rollback()
        print("Something went wrong. Please fill in the details again.")
        print("Error:", e)
        
```

```python



def Delete_Record():
    id=int(input("enter  id you want to delete record"))
    cursor.execute("select  id from B_Boys")
    ID=cursor.fetchall()

    if id==ID:
        cursor.execute("DELETE FROM B_Boys WHERE id = ?", (id,))

        conn.commit()
        print("Record deleted successfully!")


        print("After  Altering the  database table, final data is  ")

        cursor.execute("select  *from B_Boys")
        data=cursor.fetchall()
        data

        
    else:
        print("Invalid Entered ID")
            


```

```python
def Fix_DateFormat():
    
        formats = [
            "%Y-%m-%d",
            "%d-%m-%Y",
            "%Y/%m/%d",
            "%d/%m/%Y",
            "%Y-%d-%m" 
        ]
    
        cursor.execute("SELECT id, DOB FROM B_Boys")
        data = cursor.fetchall()
    
        for row in data:
            ID, DOB = row
    
            if not DOB:
                continue
    
            for fmt in formats:
                try:
                    date_obj = dt.datetime.strptime(str(DOB).strip(), fmt)
    
                    corrected_date = date_obj.strftime("%Y-%m-%d")
    
                    if DOB != corrected_date:
                        cursor.execute(
                            "UPDATE B_Boys SET DOB = ? WHERE id = ?",
                            (corrected_date, ID)
                        )
    
                        print(f"Updated: {DOB} → {corrected_date}")
    
                    break
    
                except ValueError:
                    continue
    
            else:
                print(f"Unable to fix DOB: {DOB}")
    
            conn.commit()    
            
        
```

```python

def Cleaning():
    try:
        # First, identify records containing NULL values
        # Find empty records first
        cursor.execute("""
            SELECT *
            FROM B_Boys
            WHERE name = ''
            OR id = ''
            OR DOB = ''
        """)

        data = cursor.fetchall()

        print("Records with empty entries:")

        for row in data:
            print(row)

        # Delete empty records
        cursor.execute("""
            DELETE FROM B_Boys
            WHERE name = ''
            OR id = ''
            OR DOB = ''
        """)

        print(f"\nTotal records deleted: {cursor.rowcount}")

        conn.commit()

    except sqlite3.Error as e:
        conn.rollback()
        print("Cleaning failed:", e)
```

```python
def See_details():
    cursor.execute("SELECT name, DOB, Email FROM B_Boys")
    data = cursor.fetchall()

    if not data:
        print("No records found!")
        return

    
        
    for row in data:
        print(f"Name: {row[0]}")
        print(f"DOB: {row[1]}")
        print(f"Email: {row[2]}")




print("1: Add Details")
print("2: Delete Record (using ID)")
print("3: See All Records")
print("4 : Fix Date Format ")
print("5 :Cleaning Data (null id )")
print("6: Exit")

while True:
    try:
        ch = int(input("\nEnter your choice: "))

        if ch == 1:
            Add_Details()

        elif ch == 2:
            Delete_Record()

        elif ch == 3:
            See_details()
        elif ch==4:
            Fix_DateFormat()
        elif ch==5:
            Cleaning()
            
        elif ch == 6:
            print("We're leaving the program.")
            break

        else:
            print("Invalid choice! Enter between 1 and 4.")

    except ValueError:
        print("Invalid input! Please enter a number.")

    except sqlite3.Error as e:
        conn.rollback()
        print(f"Database error: {e}")

    except Exception as e:
        print(f"Unexpected error: {e}")
        
conn = sqlite3.connect(database, timeout=60)
```
